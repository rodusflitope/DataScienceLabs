### 📌 Análisis individual [3.0 puntos]

````python
import pandas as pd

clientes_df = pd.read_parquet('data/clientes.parquet')
productos_df = pd.read_parquet('data/productos.parquet')

# Calcula el tamaño del universo total (Clientes x Productos) para el True Negative (TN)
universe_size = len(clientes_df) * len(productos_df)

def get_batch_performance(batch_id):
    # Carga de datos reales (Ground Truth) y creación de sets de pares
    actual_df = pd.read_parquet(f"data/batch_{batch_id}/batch_t{batch_id}.parquet")
    actual_pairs = set(map(tuple, actual_df[['customer_id', 'product_id']].values))
    
    # Carga de predicciones y creación de sets de pares
    preds_df = pd.read_csv(
        f"predictions/batch_{batch_id}/predictions_next_week.csv",
        header=None,
        names=['customer_id', 'product_id']
    )
    pred_pairs = set(map(tuple, preds_df[['customer_id', 'product_id']].values))
    
    # Cálculo de la matriz de confusión usando operaciones de conjuntos
    tp = len(pred_pairs & actual_pairs) # True Positives (Aciertos)
    fp = len(pred_pairs - actual_pairs) # False Positives (Ruido)
    fn = len(actual_pairs - pred_pairs) # False Negatives (Omisiones)
    # True Negatives se calcula por diferencia respecto al universo total
    tn = universe_size - tp - fp - fn
    
    predictions_size = len(pred_pairs)
    total_compras_reales = len(actual_pairs)

    # Cálculo de métricas globales
    precision = tp / predictions_size if predictions_size > 0 else 0
    recall = tp / total_compras_reales if total_compras_reales > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    accuracy = (tp + tn) / universe_size

    # Marcando si cada predicción fue un acierto o no para el análisis desagregado
    preds_df['is_hit'] = preds_df.apply(lambda x: (x['customer_id'], x['product_id']) in actual_pairs, axis=1)
    
    # Agrupación por cliente para obtener estadísticas individuales
    user_stats = preds_df.groupby('customer_id').agg(
        total_recs=('product_id', 'count'),
        hits=('is_hit', 'sum')
    )
    user_stats['user_precision'] = user_stats['hits'] / user_stats['total_recs']
    
    # Cálculo de métricas de distribución por usuario para el informe
    total_users_with_recs = len(user_stats)
    pct_precision_zero = (len(user_stats[user_stats['user_precision'] == 0]) / total_users_with_recs) * 100
    pct_precision_gt_10 = (len(user_stats[user_stats['user_precision'] > 0.10]) / total_users_with_recs) * 100
    
    metrics = {
        "global": {
            "accuracy": accuracy,
            "precision": precision,
            "recall": recall,
            "f1": f1,
            "predictions_count": predictions_size,
            "tp": tp, 
            "fp": fp, 
            "fn": fn
        },
        "user_stats": {
            "pct_users_precision_zero": pct_precision_zero,
            "pct_users_precision_gt_10": pct_precision_gt_10,
            "avg_recs_per_user": user_stats['total_recs'].mean(),
            "max_recs_per_user": user_stats['total_recs'].max()
        }
    }

    return metrics
````

Para complementar el análisis de las métricas globales, se incorporaron métricas de desempeño a nivel de cliente (user_stats). 

Este análisis se utiliza para evaluar la equidad y la robustez del modelo a través de diferentes segmentos de usuarios, ya que los promedios globales pueden ocultar fallos sistémicos para grupos específicos.

Las métricas clave agregadas y su significado son:

pct_users_precision_zero (Porcentaje de Usuarios con Precisión Cero): Indica la proporción de clientes que recibieron sugerencias pero donde ninguna de ellas resultó en una compra real (acierto). Un valor alto en esta métrica revela un grupo para el cual el modelo predice aleatoriamente o carece de información útil, lo que requiere estrategias de filtrado o modelos segmentados.

pct_users_precision_gt_10 (Porcentaje de Usuarios con Precisión Mayor al 10%): Señala el tamaño del segmento de clientes para el cual el modelo es consistente y efectivo. Este grupo representa el target ideal donde las recomendaciones funcionan bien (definiendo bien como 10% de precisión).

avg_recs_per_user y max_recs_per_user: Cuantifican la cantidad de productos máxima y promedio por cliente. Estos valores son para definir políticas de Top-N y asegurar que la cantidad de sugerencias emitida sea viable y no sature al usuario final (definidas desde el error ocurrido por el negative sampling).

````python
import pandas as pd

clientes_df = pd.read_parquet('data/clientes.parquet')
productos_df = pd.read_parquet('data/productos.parquet')
universe_size = len(clientes_df) * len(productos_df)
transacciones_df = pd.read_parquet('./data/transacciones.parquet')


print(f"Total datos: {len(transacciones_df):,} transacciones")

metrics = get_batch_performance(1)
print(f"\nMétricas Batch 1:")
for section, data in metrics.items():
    print(f"\n {section.upper()} ")
    for k, v in data.items():
          print(f"  {k}: {v:.4f}")
````

#### 📌 Batch 04 (20/01/25 - 26/01/25) [0.75 puntos]

````python
import pandas as pd

def analyze_prediction_distribution(batch_id):
    #Leer csv de predicciones
    preds_df = pd.read_csv(
        f"predictions/batch_{batch_id}/predictions_next_week.csv",
        header=None,
        names=['customer_id', 'product_id']
    )
    
    # Análisis de distribución de predicciones
    total_preds = len(preds_df)
    unique_users = preds_df['customer_id'].nunique()
    unique_products = preds_df['product_id'].nunique()
    
    user_stats = preds_df.groupby('customer_id').size()
    
    # Cálculo de estadísticas
    stats = {
        "volume": {
            "total_predictions": total_preds,
            "unique_users_targeted": unique_users,
            "unique_products_recommended": unique_products
        },
        "distribution": {
            "avg_recs_per_user": user_stats.mean(),
            "median_recs_per_user": user_stats.median(),
            "min_recs_per_user": user_stats.min(),
            "max_recs_per_user": user_stats.max(),
            "std_recs_per_user": user_stats.std()
        }
    }
    return stats

stats = analyze_prediction_distribution(4)

for section, data in stats.items():
    print(f"\n--- {section.upper()} ---")
    for k, v in data.items():
        if isinstance(v, float):
             print(f"  {k:<30}: {v:.4f}")
        else:
             print(f"  {k:<30}: {v:,}")
````

La distribución de las predicciones del Batch 4 replica casi con exactitud el patrón del Batch 3 (que tuvo un promedio de 9,60). Esto indica que el modelo mantiene una estrategia de alta cobertura, priorizando el volumen sobre la selectividad observada en el Batch 2. Dado que este volumen conlleva un mayor riesgo de fatiga del cliente (saturación por exceso de oferta), se recomienda aplicar filtros estrictos (Top-N) antes del envío final si se desea priorizar la experiencia de usuario sobre la cobertura bruta.

### 📌 Análisis comparativo [2.0 puntos]

````python
import pandas as pd
import os

BATCH_IDS = [1, 2, 3, 4]

def get_comparative_stats(batch_ids):
    
    report_data = []

    # Cargar Baseline (Predicciones del Batch 1)
    base_path = "predictions/batch_1/predictions_next_week.csv"
    base_df = pd.read_csv(base_path, header=None, names=['customer_id', 'product_id'])
    base_pairs = set(map(tuple, base_df[['customer_id', 'product_id']].values))
    base_count = len(base_pairs)

    for b_id in batch_ids:
        
        # Cargar Predicciones del Batch Actual
        pred_path = f"predictions/batch_{b_id}/predictions_next_week.csv"
            
        preds_df = pd.read_csv(pred_path, header=None, names=['customer_id', 'product_id'])
        pred_pairs = set(map(tuple, preds_df[['customer_id', 'product_id']].values))
        predictions_count = len(pred_pairs)
        
        # Cargar Ground Truth
        true_path = f"data/batch_{b_id}/batch_t{b_id}.parquet"
        has_truth = os.path.exists(true_path)
        
        metrics = {}
        
        # INFO GENERAL
        metrics['Batch'] = f"Batch {b_id}"
        metrics['Predicciones'] = predictions_count
        
        if has_truth:
            actual_df = pd.read_parquet(true_path)
            actual_pairs = set(map(tuple, actual_df[['customer_id', 'product_id']].values))
            total_real = len(actual_pairs)
            
            # RENDIMIENTO MODELO ACTUAL
            tp = len(pred_pairs & actual_pairs)
            precision = tp / predictions_count if predictions_count > 0 else 0
            recall = tp / total_real if total_real > 0 else 0
            f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
            
            # RENDIMIENTO BASELINE (PREDICCIONES BATCH 1)
            tp_base = len(base_pairs & actual_pairs)
            prec_base = tp_base / base_count if base_count > 0 else 0
            rec_base = tp_base / total_real if total_real > 0 else 0
            f1_base = 2 * (prec_base * rec_base) / (prec_base + rec_base) if (prec_base + rec_base) > 0 else 0

            # User stats
            user_counts = preds_df.groupby('customer_id').size()
            preds_df['is_hit'] = preds_df.apply(lambda x: (x['customer_id'], x['product_id']) in actual_pairs, axis=1)
            user_hits = preds_df.groupby('customer_id')['is_hit'].sum()
            user_precision = user_hits / user_counts
            pct_zero = (len(user_precision[user_precision == 0]) / len(user_precision)) * 100

            metrics['F1 Actual'] = f1
            metrics['F1 Baseline (B1)'] = f1_base
            metrics['Delta F1'] = f1 - f1_base
            metrics['Precision Actual'] = precision
            metrics['Precision Base'] = prec_base
            metrics['% Usr Prec 0'] = pct_zero
            
        else:
            # Sin datos reales (Batch 4)
            metrics['F1 Actual'] = None
            metrics['F1 Baseline (B1)'] = None
            metrics['Delta F1'] = None
            metrics['Precision Actual'] = None
            metrics['Precision Base'] = None
            metrics['% Usr Prec 0'] = None

        report_data.append(metrics)

    return pd.DataFrame(report_data)

# Ejecutar análisis
df_comparativo = get_comparative_stats(BATCH_IDS)

# Visualización
cols_show = [
    'Batch', 
    'F1 Actual', 'F1 Baseline (B1)', 'Delta F1',
    'Precision Actual', 'Precision Base', 
    '% Usr Prec 0'
]

print(df_comparativo[cols_show].to_string(index=False, float_format="%.4f"))
````

El análisis comparativo revela un comportamiento contraintuitivo, las predicciones generadas por el modelo original (usadas para el Batch 1) superaron consistentemente en rendimiento a las versiones reentrenadas. Se observa un Delta F1 negativo progresivo, comenzando con una leve disminución de -0,0062 en el Batch 2 y acentuándose drásticamente en el Batch 3 con una caída de -0,0861 (donde el modelo base mantuvo un F1 de 0,46 frente al 0,37 del actual). Este deterioro en las métricas sugiere que la estrategia de reentrenamiento incremental podría estar introduciendo ruido o sobreajuste en lugar de aprendizaje útil, o que se implemento de manera incorrecta el reentrenamiento a nivel de código.

### 📌 Conclusiones y Aprendizajes [1.0 puntos]

La implementación de la optimización de hiperparámetros para calibrar el scale_pos_weight fue el mayor avance a nivel de rendimiento en f1, logrando estabilizar el desbalance de clases.

Sin embargo, ocurrió un problema al momento de agregar cada batch, la degradación por reentrenamiento continuo. Los datos demuestran que el modelo original (Batch 1) generalizó mejor en periodos futuros que sus versiones actualizadas, las cuales mostraron una tendencia a priorizar la cobertura desmedida (sacrificando precisión) a medida que ingerían más datos de nuevos periodos.

Como lección principal, se concluye que el volumen de datos no garantiza calidad predictiva. Para las siguientes iteraciones, es imperativo auditar la estrategia de ingestión de datos y priorizar el uso del modelo base estabilizado, además es importante la revisión del código de forma exhaustiva para comprobar que el motivo de esto no sea la forma en que estos nuevos datos se usaron.
